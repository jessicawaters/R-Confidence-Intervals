# Confidence Intervals

## Simple computations

### Example 1

If the error probability is $0.05$ (and the confidence level is $0.95$) and we are working with the sample mean, then

```r
z_critical <- qnorm(1-0.05/2)
z_critical
```

If the sample mean is 50 and the sample standard deviation is 10, and the sample size is 40, then the confidence interval is...

```r
xbar <- 50
s <- 10
n <- 40

lower <- xbar - z_critical * s / sqrt(n)
upper <- xbar + z_critical * s / sqrt(n)

print(lower)
print(upper)
```

### Example 2

If the sample proportion is 0.3 and the sample size is 40, then the 95% confidence interval is

```r
p_hat <- 0.3
n <- 40
s <- sqrt(p_hat * (1 - p_hat) / n)
z_critical <- qnorm(1 - 0.05 / 2)
lower_bound <- p_hat - z_critical * s
upper_bound <- p_hat + z_critical * s
print(lower_bound)
print(upper_bound)
```

### Confidence interval on mean weight

Let's sample 100 dogs from the data and compute the mean weight of the sample.  We'll use that to construct a 95% confidence interval and compare to the mean weight of all the dogs in the data set.

```r
set.seed(1234)
sample_data <- dogs[sample(nrow(dogs), 100), "Weight..kg."]
sample_mean <- mean(sample_data)
sample_sd <- sd(sample_data)
sample_size <- length(sample_data)
z_critical <- qnorm(1 - 0.05 / 2)
lower_bound <- sample_mean - z_critical * sample_sd / sqrt(sample_size)
upper_bound <- sample_mean + z_critical * sample_sd / sqrt(sample_size)
print(lower_bound)
print(upper_bound)
```

The mean weight of all dogs in the data set is

```r
mean(dogs$Weight..kg.)
```

You can also use the `t.test` function.  This will give a confidence interval that is just a teensy bit wider.

```r
t.test(sample_data,conf.level=0.95)
```

### Confidence interval on proportion of female dogs

```r
set.seed(1234)
sample_data <- dogs[sample(nrow(dogs), 100), "Gender"]
sample_proportion <- mean(sample_data == "Female")
sample_sd <- sqrt(sample_proportion * (1 - sample_proportion))
sample_size <- length(sample_data)
z_critical <- qnorm(1 - 0.05 / 2)
lower_bound <- sample_proportion - z_critical * sample_sd / sqrt(sample_size)
upper_bound <- sample_proportion + z_critical * sample_sd / sqrt(sample_size)
print(lower_bound)
print(upper_bound)
```

```r
mean(dogs$Gender == "Female")
```
